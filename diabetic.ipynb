```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statistics as stat
import seaborn as sns
import plotly.express as px
from bokeh.plotting import figure, show
```

```python
pd.set_option('display.max_columns', None)
db = pd.read_csv('diabetic.csv')
db_copy = db.copy()
db.head()
```

```python
db.duplicated().sum()
```

```python
db.isnull().sum()
```

```python
# Count how many times '?' appears in each column
missing_counts = (db == '?').sum()
# Only show columns that actually have missing values
print(missing_counts[missing_counts > 0])
```

```python
# replacing "?" with null value
db.replace('?', np.nan, inplace=True)
db.head()
```

```python
missing_values = db.isnull().sum()
print(missing_values[missing_values > 0])
```

### Data Standarization 

```python
# Fill the missing values in the medical_specialty column
db['medical_specialty'] = db['medical_specialty'].fillna('Unknown')
```

```python
# Standardize each column individually in the db dataframe
db['race'] = db['race'].replace('?', 'Unknown').fillna('Unknown')
db['payer_code'] = db['payer_code'].replace('?', 'Unknown').fillna('Unknown')
db['admission_type'] = db['admission_type'].replace('?', 'Unknown').fillna('Unknown')
db['discharge_disposition'] = db['discharge_disposition'].replace('?', 'Unknown').fillna('Unknown')
db['admission_source'] = db['admission_source'].replace('?', 'Unknown').fillna('Unknown')
```

```python
db_final = db.fillna('Null')
db_final.head()
```

```python
# Save the final version
# The 'utf-8-sig' encoding is the secret to making CSVs look perfect in Excel
db_final.to_csv('diabetic_final_cleaned.csv', index=False, encoding='utf-8-sig')
```
